## 1. Predice la evaluación paso a paso

Dada la expresión:

```haskell
take 5 (map (*2) [1..])
```

### a. Predicción del resultado sin ejecutarlo

El resultado esperado es:

```haskell
[2,4,6,8,10]
```

### Evaluación paso a paso

```haskell
take 5 (map (*2) [1..])
= take 5 (map (*2) (1:[2..]))
= take 5 ((2) : map (*2) [2..])
= 2 : take 4 (map (*2) [2..])
= 2 : 4 : take 3 (map (*2) [3..])
= 2 : 4 : 6 : take 2 (map (*2) [4..])
= 2 : 4 : 6 : 8 : take 1 (map (*2) [5..])
= 2 : 4 : 6 : 8 : 10 : take 0 (map (*2) [6..])
= [2,4,6,8,10]
```

### b. Ejecútalo y compara tu predicción

````haskell
take 5 (map (*2) [1..])
````

## 2. Pregunta de análisis

### ¿Por qué `[1..]` no causa un ciclo infinito?

`[1..]` es una lista infinita, pero Haskell usa evaluación perezosa.  
Eso significa que no intenta construir la lista completa desde el inicio, sino que genera elementos solo cuando otra función los necesita.

En este caso, take 5 solo pide los primeros cinco valores.  
Por eso Haskell calcula únicamente:

`1, 2, 3, 4, 5`

Luego les aplica `(*2)` y se detiene.

## 3. Considera la siguiente función

````haskell
misterio :: Int -> Int -> Int
misterio x y = if x > 10 then y * 2 else x + y
````

### a. Evalúa paso a paso `misterio (5+7) (4*3)`

```haskell
misterio (5+7) (4*3)
= if (5+7) > 10 then (4*3) * 2 else (5+7) + (4*3)
= if 12 > 10 then 12 * 2 else (5+7) + (4*3)
= 12 * 2
= 24
```
