## 1. Define las siguientes funciones

### a. `sumaCuadrados :: Int -> Int -> Int`
Que calcule la suma de los cuadrados de dos números, definiendo también `suma` y `cuadrado`.

````haskell
suma :: Int -> Int -> Int
suma x y = x + y

cuadrado :: Int -> Int
cuadrado x = x * x

sumaCuadrados :: Int -> Int -> Int
sumaCuadrados x y = suma (cuadrado x) (cuadrado y)

sumaCuadrados 3 4
````

### b. `dobleLista :: [Int] -> [Int]`
Multiplica por 2 cada elemento de una lista usando `map` y una función `duplicar`.

````haskell
duplicar :: Int -> Int
duplicar x = x * 2

dobleLista :: [Int] -> [Int]
dobleLista xs = map duplicar xs

dobleLista [1,2,3,4]
````

## 2. Clasificación de notas

### a. `clasificarNota :: Int -> String`

Debe retornar:

- Menor a 3 → `"Reprobado"`
- Entre 3 y 4 → `"En recuperación"`
- Mayor o igual a 4 → `"Aprobado"`

````haskell
clasificarNota :: Int -> String
clasificarNota n
  | n < 3     = "Reprobado"
  | n < 4     = "En recuperación"
  | otherwise = "Aprobado"

clasificarNota 2
clasificarNota 3
clasificarNota 5
````

## 3. Fibonacci

### b. `fibonacci :: Int -> Int`

Se implementa de forma recursiva.

````haskell
fibonacci :: Int -> Int
fibonacci 0 = 0
fibonacci 1 = 1
fibonacci n
  | n > 1     = fibonacci (n - 1) + fibonacci (n - 2)
  | otherwise = error "Fibonacci no está definido para negativos"

fibonacci 0
fibonacci 1
fibonacci 6
````
